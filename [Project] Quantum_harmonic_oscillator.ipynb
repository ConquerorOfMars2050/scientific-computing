```python
E=0
psi=1
counter=1
xvalues=[]
eigenfunctions = []
eigenenergies = []
nmax = 10

while counter<=nmax:
    psi=1
    
    while abs(psi)>0.02:
        psi=0.01
        dpsi=0.0
        E=E+0.05
        a=np.sqrt(10+E/V)
        x=-a
        xlist=[]
        psilist=[]
        
    
        while x<=a:
            ddpsi=2*m/h**2*(V*x**2-E)*psi
            dpsi=dpsi+ddpsi*dx
            psi=psi+dpsi*dx
            x=x+dx
            xlist.append(x)
            psilist.append(psi)
            
    eigenfunctions.append([psilist])
    eigenenergies.append([E])
    xvalues.append(xlist)
    counter = counter+1
    E=E*1.1
```

```python
counter=1
while counter<=nmax:
    plt.plot(xvalues[counter-1], eigenfunctions[counter-1][0])
    counter=counter+1
```

```python
counter=1
while counter<=nmax:
    norm=np.dot(eigenfunctions[counter-1][0], eigenfunctions[counter-1][0])*dx
    eigenfunctions[counter-1][0][:] = [x / np.sqrt(norm) for x in eigenfunctions[counter-1][0]]
    counter = counter+1
```
